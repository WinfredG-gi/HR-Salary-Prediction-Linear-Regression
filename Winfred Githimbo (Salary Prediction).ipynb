```python
# data preprocessing: convert text in experience to numbers first then replace missing values
experience_map = {
    'two': 2,
    'three': 3,
    'five': 5,
    'seven': 7,
    'ten': 10,
    'eleven': 11
}
```

```python
df['experience'] = df['experience'].map(experience_map)
```

```python
#Replace missing values in experience column with 0
df['experience'].fillna(0, inplace=True)
```

```python
#Replace missing value in test score with median value
df['test_score(out of 10)'].median()
```

```python
df['test_score(out of 10)'] = df['test_score(out of 10)'].fillna(df['test_score(out of 10)'].median())
df
```

```python
df['experience'] = df['experience'].astype(int)
```
